```python
import pandas as pd

file_path = "/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv"

data = pd.read_csv(file_path)

print("Dataset shape:", data.shape)

print("\nColumn names:")
print(data.columns.tolist())

print("\nFirst 5 rows:")
display(data.head())
```

```python
# Check search performance and content decline signals

print("Total content records:", len(data))
print("Unique clients:", data["client_id"].nunique())

print("\nAvailable relevant columns:")
print([
    col for col in data.columns
    if any(word in col.lower() for word in
           ["impression", "position", "volatility", "click", "refresh"])
])
```

```python
# Show basic statistics for relevant numeric columns

keywords = ["impression", "position", "volatility", "click"]

for col in data.columns:
    if any(word in col.lower() for word in keywords):
        if pd.api.types.is_numeric_dtype(data[col]):
            print(f"\n{col}")
            print(data[col].describe())
```

```python
print("Average impressions:", data["impressions_last_30d"].mean())
print("Average clicks:", data["clicks_last_30d"].mean())
print("Average position:", data["avg_position"].mean())
```

```python
print("Available columns:")
print(data.columns.tolist())
```

```python
for col in data.columns:
    if any(word in col.lower() for word in ["impression", "position", "volatility", "click", "date", "week"]):
        print(col)
```

```python
# Compare impressions in the previous and last 30 days

data["impression_change"] = (
    data["impressions_last_30d"] -
    data["impressions_prev_30d"]
)

data["impression_change_pct"] = (
    data["impression_change"] /
    data["impressions_prev_30d"].replace(0, float("nan")) * 100
)

print("Average impressions in previous 30 days:",
      data["impressions_prev_30d"].mean())

print("Average impressions in last 30 days:",
      data["impressions_last_30d"].mean())

print("Pages with declining impressions:",
      (data["impression_change"] < 0).sum())

print("Pages with increasing impressions:",
      (data["impression_change"] > 0).sum())
```
